```python
#Section 2 - No.5 IF Statement and Loops

# one condition
if (0 < 1):

    print('Yes')
```

```python
# two conditions
if (9 < 3):

    print('Yes')

else:

    print('No')
```

```python
# multiple conditions

if (6 < 2):

    print(1)

elif (3 == 3):

    print(2)

elif (4 == 4):

    print(3)

else:

    print(4)
```
