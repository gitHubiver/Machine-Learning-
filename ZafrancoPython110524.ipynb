**Activity 2 : Line Graph Customization**

```python
#Module Sample

# Basic line graph
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
plt.plot(x, y, marker='o') # 'marker' specifies the points on the line
plt.title("Line Graph Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()
```

```python
#Line Grapgh with custom label and title

fig = plt.figure(figsize = (5, 5))
fig.suptitle('Electives Title') # This is a figure title, not the axes title
plt.title('Axes Title') # This is the axes title

plt.xlabel('X Axis Label') # X axis label
plt.ylabel('Y Axis Label') # Y axis label

plt.plot([1, 2, 3], [4, 5, 6])
plt.show()
```

```python
#Experiment on different styling of Line Graph

fig = plt.figure(figsize=(5, 5))
fig.suptitle('Electives Figure Title')  # Figure title
ax = fig.add_subplot(1, 1, 1)  # Add axes
ax.set_title('Axes Title')      # Axes title

# Customize the x and y axis labels
ax.set_xlabel('X Axis Label')
ax.set_ylabel('Y Axis Label')

# Add grid, customized line style, color, and red dashed line
ax.grid(True, linestyle='--', color='gray', alpha=0.7)  # Adding a grid with dashed gray lines
ax.plot([1, 2, 3], [4, 5, 6], color='red', linestyle='--', linewidth=2)  # Red dashed line with thicker line width

plt.show()
```

```python
#Experiment on different styling of Line Graph

# Sample data representing sales over three months for two products
months = ['January', 'February', 'March']
product_A_sales = [150, 200, 250]
product_B_sales = [180, 170, 210]

# Set up the figure and axes with titles
fig = plt.figure(figsize=(7, 5))
fig.suptitle('Monthly Sales Comparison')  # Figure title
ax = fig.add_subplot(1, 1, 1)             # Add axes
ax.set_title('Product Sales Over Three Months')  # Axes title

ax.set_xlabel('Months')
ax.set_ylabel('Sales')

ax.grid(True, linestyle='--', color='gray', alpha=0.7)  # Adding a grid with dashed gray lines

# Plot data for Product A with custom style
ax.plot(months, product_A_sales, color='red', linestyle='--', marker='o', linewidth=2, label='Product A')

# Plot data for Product B with a different style
ax.plot(months, product_B_sales, color='blue', linestyle='-.', marker='s', linewidth=2, label='Product B')

ax.legend()
plt.show()
```

**Activity 3 : Customizing Bar Graphs**

```python
#Module Sample

# Basic bar graph
categories = ['A', 'B', 'C', 'D']
values = [10, 20, 15, 30]
plt.bar(categories, values, color="teal")
plt.title("Bar Graph Example")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
#Customized bars with different colors each bar

categories = ['A', 'B', 'C', 'D']
values = [10, 20, 15, 30]
colors = ['teal', 'orange', 'purple', 'green']

plt.bar(categories, values, color=colors)
plt.title("Bar Graph Example")

plt.show()
```

```python
#With Labels and Grid Lines

categories = ['A', 'B', 'C', 'D']
values = [10, 20, 15, 30]
colors = ['teal', 'orange', 'purple', 'green']  # List of colors for each bar

plt.figure(figsize=(7, 5))
plt.bar(categories, values, color=colors)  # Apply different colors to each bar

# Add titles and labels
plt.title("Bar Graph Example")
plt.xlabel("Categories")
plt.ylabel("Values")

# Add grid lines
plt.grid(axis='y', linestyle='--', color='gray', alpha=0.7)  # Horizontal grid lines

# Add labels on top of each bar
for i, value in enumerate(values):
    plt.text(i, value + 1, str(value), ha='center', va='bottom', fontweight='bold')  # Labels above each bar

plt.show()
```

**Activity 4 :  Creating Subplots**

```python
#Module Sample
# Using subplots
fig, axs = plt.subplots(1, 2, figsize=(10,4)) #1 row, 2 columns
# Line plot in the first subplot axs[0].plot(x, y, color='blue')
axs[0].set_title("Line Graph")
# Bar plot in the second subplot
axs[1].bar(categories, values, color='purple')
axs[1].set_title("Bar Graph")
plt.tight_layout()
plt.show()
```

```python
#Subplot with line graph in the first row  and  bar graph

games = np.arange(1, 11)  # 10 games
points = np.random.randint(80, 130, size=10)  # Random points between 80 and 130

# Create subplots: 2 rows, 1 column
fig, axs = plt.subplots(2, 1, figsize=(8, 8))

# Line graph in the first row
axs[0].plot(games, points, marker='o', color='b', linestyle='-', markersize=8)
axs[0].set_title('Basketball Points: Line Graph')
axs[0].set_xlabel('Game')
axs[0].set_ylabel('Points')
axs[0].grid(True)

# Bar chart in the second row
axs[1].bar(games, points, color='g')
axs[1].set_title('Basketball Points: Bar Chart')
axs[1].set_xlabel('Game')
axs[1].set_ylabel('Points')
axs[1].grid(True)

# Adjust layout to prevent overlap
plt.tight_layout()

# Show the plot
plt.show()
```

```python
#Module Sample

x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]  # Example data for prime numbers

# Plotting the line graph with enhancements
plt.plot(x, y, label="Prime Numbers", color='green')  # Line with label and color
plt.title("Enhanced Line Graph with Legend")           # Graph title
plt.xlabel("X-axis")                                   # X-axis label
plt.ylabel("Y-axis")                                   # Y-axis label
plt.legend()                                           # Adding a legend
plt.grid(True)                                         # Adding grid lines
plt.show()
```

```python
#With Legend and Grid as well annotated highest data point

months = ['January', 'February', 'March']
product_A_sales = [150, 200, 250]
product_B_sales = [180, 170, 210]

# Set width and positions for the bars
bar_width = 0.35
x = np.arange(len(months))

# Set up the figure and axes with titles
fig, ax = plt.subplots(figsize=(7, 5))
fig.suptitle('Monthly Sales Comparison')  # Figure title
ax.set_title('Product Sales Over Three Months')  # Axes title

# Customize the x and y axis labels
ax.set_xlabel('Months')
ax.set_ylabel('Sales (in units)')
ax.set_xticks(x)
ax.set_xticklabels(months)

# Plot the bars for Product A with a custom color for each month
bar1 = ax.bar(x - bar_width / 2, product_A_sales, width=bar_width, color=['#FF6666', '#FF3333', '#CC0000'], label='Product A')

# Plot the bars for Product B with a different custom color for each month
bar2 = ax.bar(x + bar_width / 2, product_B_sales, width=bar_width, color=['#6699FF', '#3366FF', '#0033CC'], label='Product B')

# Add grid lines
ax.grid(axis='y', linestyle='--', color='gray', alpha=0.7)  # Horizontal grid lines

# Add labels on top of each bar for Product A and Product B
for bar in bar1:
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width() / 2, height + 5, f'{height}', ha='center', va='bottom', fontweight='bold')

for bar in bar2:
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width() / 2, height + 5, f'{height}', ha='center', va='bottom', fontweight='bold')

# Find the highest data point and annotate it
max_sales_A = max(product_A_sales)
max_sales_B = max(product_B_sales)
max_sales = max(max_sales_A, max_sales_B)

# Determine the month and product of the highest sales
if max_sales == max_sales_A:
    month_index = product_A_sales.index(max_sales)
    ax.annotate('Highest Sales',
                xy=(month_index - bar_width / 2, max_sales),
                xytext=(month_index - bar_width / 2, max_sales + 30),
                arrowprops=dict(facecolor='black', arrowstyle='->'),
                ha='center', va='bottom', fontsize=10, color='black')
else:
    month_index = product_B_sales.index(max_sales)
    ax.annotate('Highest Sales',
                xy=(month_index + bar_width / 2, max_sales),
                xytext=(month_index + bar_width / 2, max_sales + 30),
                arrowprops=dict(facecolor='black', arrowstyle='->'),
                ha='center', va='bottom', fontsize=10, color='black')

# Add legend to differentiate the products
ax.legend()

plt.show()



```

Discussion Questions

1. When would you choose a bar graph over a line graph, and why?
2. What are some advantages of using subplots in a single figure?
3. How does customizing graphs improve the clarity of data visualization?

1. A bar graph is preferred over a line graph when comparing distinct categories or groups instead of showing trends over time. For example, if you want to illustrate the number of students in different programs at a university, a bar graph makes it easy to see the differences between each program. Bar graphs visually represent each category’s value clearly, allowing viewers to interpret and compare the data quickly. In contrast, line graphs are better for showing changes and trends in continuous data, like temperature over time.

2. Subplots in a single figure have several benefits, especially for related data. They allow for easy side-by-side comparisons of different datasets or aspects of the same dataset, helping viewers spot similarities or differences quickly. This layout saves space and keeps the figure organized, allowing for various graph types (like bars and lines) to be displayed together. This arrangement makes it easier to tell a story with the data and helps the audience understand complex information more clearly.

3. Customizing graphs enhances data visualization clarity by highlighting important information and improving understanding. Changes like adjusting colors, fonts, and labels make key data points stand out, directing viewers' attention where needed. Clear titles, labels, and legends eliminate confusion, ensuring the audience grasps the data easily. Consistent styles also align with branding and make the visuals more engaging. Overall, customization transforms raw data into clear visuals that effectively communicate the intended message.
