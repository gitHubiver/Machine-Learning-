```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
import requests
from bs4 import BeautifulSoup
warnings.filterwarnings('ignore')
```

Web Scrape NBA Standings from CBS Sports

```python
#URL
url = 'https://www.cbssports.com/nba/standings/'

#Send Request
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/134.0.0.0 Safari/537.36'}


# Response

response = requests.get(url = url, headers = headers)

#Create a bs4 object to parse the html content
nbaTbls = BeautifulSoup(response.content, 'html.parser')
```

Eastern Conference

```python
#Find the tbody containing east conference teams, you can use select ('tbody')[0],means first tbody found
eastConf = nbaTbls.select('tbody')[0]
```

```python
# Find the table containing teams
#table_teams = eastConf.select_one('col.TableBase-colGroupCol--team tbody')
# Find the table with the stats
#table_data = eastConf.select_one('colgroup.TableBase-colGroup')
#table_data
#table_teams
```

```python
#Find all Team Names, you can use the .get_text().strip to get the text and strip extra spaces

team = []

for teams in eastConf.find_all('span', class_ = 'TeamName'):

  team.append(teams.get_text())

#team
```

```python
#Data  for Eastern Conference Teams

data = []
for tr in eastConf.find_all('tr'):
  row_data = []
  for i, td in enumerate(tr.find_all('td')):
    if i > 1:
      row_data.append(td.get_text().strip())
  data.append(row_data)

#data
```

```python
#Construct Data Frame of Eastern Coference Teams

team_df = pd.DataFrame(team, columns = ['Team'])
team_df
```

```python
#Column Names

col_names = ['W','L','PCT','GB','PPG','OPPG','Diff','Home','Road','Div','Conf','STRK','L10','W','Div','Post']
#col_names
print(len(data[0]))
```

```python
#Construct Data Frame for Eastern Conference Teams

team_data = pd.DataFrame(data, columns = col_names)
team_data
```

```python
#Combine both DataFrame (Eastern Conference)

df_final = pd.concat([team_df, team_data], axis = 1)
df_final
```

Western Conference

```python
#Find the tbody containing west conference teams, you can use select ('tbody')[0],means second tbody found
westConf = nbaTbls.select('tbody')[1]
```

```python
#Team Names Western Conference
team = []

for teams in westConf.find_all('span', class_ = 'TeamName'):

  team.append(teams.get_text())

#team
```

```python
#Data for Western Conference


data = []
for tr in westConf.find_all('tr'):
  row_data = []
  for i, td in enumerate(tr.find_all('td')):
    if i > 1:
      row_data.append(td.get_text().strip())
  data.append(row_data)

#data
```

```python
#Teams Data Frame for Western Conference

team_df2 = pd.DataFrame(team, columns = ['Team'])
team_df2
```

```python
#Data Frame of Western Coference

team_data2 = pd.DataFrame(data, columns = col_names)
team_data2
```

```python
#Combine both DataFrame (Western Conference)

df_final2 = pd.concat([team_df2, team_data2], axis = 1)
df_final2
```

Combine Eastern Conference and Western Conference Data

```python
#Combine both DataFrame (Eastern Conference)

df_combine = pd.concat([df_final, df_final2], axis = 0)
df_combine
```

```python
#Sort by PCT in Descending Order
df_combine.sort_values(by = 'PCT', ascending = False)
df_combine
```

```python
#Reset Index
df_combine.reset_index(drop = True, inplace = True)
df_combine
```
