```python
X = pd.get_dummies(df.drop('Species', axis = 1), drop_first = True)
y = df['Species']
```

```python
# Train - Test Split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 50)
```

```python
# Instance of Model

from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()
model.fit(X_train, y_train)
```

```python
base_preds = model.predict(X_test)
base_preds
```

```python

# Classification Report

from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(y_test, base_preds))
```

```python
cm = confusion_matrix(y_test, base_preds)
sns.heatmap(cm, annot = True, fmt = 'g', xticklabels = ['Iris-setosa', 'Iris-versicolor', 'Iris-virginica'], yticklabels = ['Iris-setosa', 'Iris-versicolor', 'Iris-virginica']).set(title = 'Confusion Matrix')
plt.show()
```

```python
# Some Decision Tree Model Attributes

model.feature_importances_
pd.DataFrame(index = X.columns, columns = ['Feature Importance'], data = model.feature_importances_).sort_values('Feature Importance', ascending = False)
```

```python
# Tree Visualization

from sklearn.tree import plot_tree # Implementation, left (true), right (false)

plt.figure(figsize = (30,10), dpi = 200)
plot_tree(model, feature_names = X.columns, filled = True);
```

```python
tree_structure = model.tree_

# The number of leaves is stored in the 'n_leaves' attribute
num_terminal_nodes = tree_structure.n_leaves

print("Number of terminal nodes:", num_terminal_nodes)
```

```python
report = classification_report(y_test, base_preds, output_dict=True)
accuracy = report['accuracy']
print(f"Accuracy of the Decision Tree model: {accuracy}")
```
