```python
validation_data = [[3.15, 3.2, 1.21, 0.2, 'Unknown'],
                  [1.35, 4.1, 1.7, 0.2, 'Unknown']]

validation_data = pd.DataFrame(validation_data, columns = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm', 'Species'])
validation_data
```

```python
df = pd.concat([df, validation_data], axis = 0)
df.replace(['Iris-setosa', 'Iris-versicolor', 'Iris-virginica','Unknown'], [0,1,2,-1], inplace = True)
```

```python
df.reset_index(drop = True, inplace = True)
df.tail(5)
```

```python
X = df.drop('Species', axis = 1)

scaler = StandardScaler()
scaler.fit(X)
```

```python
scaled_features_data = scaler.transform(X)
scaled_features_data
```

```python
scaled_features_data = pd.DataFrame(scaled_features_data, columns = df.columns[:-1]) # all columns except 'Species'
scaled_features_data.tail()
```

```python
X = scaled_features_data.iloc[:158,:]
y = df.iloc[:158, -1]
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 101)
```

```python
X_train
```

```python
y_train
```

```python
model = SVC(kernel = 'linear', C = 1)
model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)
predictions
```

```python
cm = confusion_matrix(y_test, predictions)
sns.heatmap(cm, annot = True, fmt = 'd')
```

```python
print(accuracy_score(y_test, predictions))
```

```python

print(classification_report(y_test, predictions))
```

```python
from sklearn.model_selection import GridSearchCV

svm = SVC()
param_grid = {'C': [0.01, 0.05, 0.1, 0.5, 1, 1.5, 3, 5, 10, 25, 50, 100], 'kernel': ['linear', 'rbf', 'poly']}
grid = GridSearchCV(svm, param_grid)
grid.fit(X, y)
```

```python
grid.best_params_
```
